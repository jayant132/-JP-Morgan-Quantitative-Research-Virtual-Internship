```python
import pandas as pd

df = pd.read_csv("Task 3 and 4_Loan_Data.csv")
```

```python
print(df.head())
```

```python
print(df.columns)
```

```python
df['fico_bucket'] = pd.qcut(
    df['fico_score'],
    q=5,
    labels=False
)
```

```python
df['rating'] = 5 - df['fico_bucket']
```

```python
print(df[['fico_score', 'fico_bucket', 'rating']].head(10))
```

```python
default_rates = df.groupby('rating')['default'].mean()

print(default_rates)
```

```python
bucket_ranges = df.groupby('rating')['fico_score'].agg(['min', 'max'])

print(bucket_ranges)
```
