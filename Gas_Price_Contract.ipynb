```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import interp1d
```

```python
df = pd.read_csv("Nat_Gas.csv")
```

```python
df['Dates'] = pd.to_datetime(df['Dates'])
```

```python
df['Days'] = (df['Dates'] - df['Dates'].min()).dt.days
```

```python
interp_model = interp1d(
    df['Days'],
    df['Prices'],
    kind='cubic',
    fill_value='extrapolate'
)
```

```python
def estimate_price(date):

    date = pd.to_datetime(date)

    days = (date - df['Dates'].min()).days

    estimated_price = interp_model(days)

    return float(estimated_price)
```

```python
print(estimate_price("2025-01-15"))
```

```python
def price_gas_contract(
    injection_date,
    withdrawal_date,
    volume,
    storage_cost_per_month
):

    buy_price = estimate_price(injection_date)

    sell_price = estimate_price(withdrawal_date)

    injection_date = pd.to_datetime(injection_date)

    withdrawal_date = pd.to_datetime(withdrawal_date)


    days_stored = (withdrawal_date - injection_date).days

    months_stored = days_stored / 30


    gross_profit = (sell_price - buy_price) * volume


    total_storage_cost = storage_cost_per_month * months_stored


    contract_value = gross_profit - total_storage_cost


    return contract_value
```

```python
value = price_gas_contract(
    injection_date="2024-06-01",
    withdrawal_date="2024-12-01",
    volume=1000000,
    storage_cost_per_month=100000
)

print(value)
```
